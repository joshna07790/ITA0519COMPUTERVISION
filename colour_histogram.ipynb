```python
import cv2
import matplotlib.pyplot as plt
filename = list(uploaded.keys())[0]

image = cv2.imread(filename)

print("Image uploaded successfully:", filename)
```

```python

image = cv2.imread(filename)

print("Image uploaded successfully:", filename)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
histogram_b = cv2.calcHist([image], [0], None, [256], [0, 256])
histogram_g = cv2.calcHist([image], [1], None, [256], [0, 256])
histogram_r = cv2.calcHist([image], [2], None, [256], [0, 256])

plt.plot(histogram_b, color='b')
plt.plot(histogram_g, color='g')
plt.plot(histogram_r, color='r')

plt.title("Color Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.show()
```
